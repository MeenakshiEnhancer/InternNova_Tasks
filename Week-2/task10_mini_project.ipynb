```python
df=pd.read_csv('retail_store_sales.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Item']=df.groupby('Category')['Item'].transform(lambda x:x.fillna(x.mode()[0]))
```

```python
df['Price Per Unit']=df['Price Per Unit'].fillna(df['Price Per Unit'].median())
```

```python
df['Quantity']=df['Quantity'].fillna(df['Quantity'].fillna(df['Quantity'].median()))
```

```python
df['Total Spent']=df['Total Spent'].fillna(df['Total Spent'].median())
```

```python
df['Discount Applied']=df['Discount Applied'].fillna(df['Discount Applied'].mode()[0])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
high_sales = df[df['Total Spent'] > 300]

high_sales.head()
```

```python
df.sort_values('Total Spent', ascending=False).head(10)
```

```python
category_sales = df.groupby('Category')['Total Spent'].sum()

category_sales
```

```python
df.head(3)
```

```python
pivot_table = pd.pivot_table(
    df,
    values='Total Spent',
    index='Payment Method',
    columns='Category',
    aggfunc='sum'
)

pivot_table
```

**Key Insights**
1)Cash generated the highest sales for Butchers, with total sales of 75,350.5.

2)Digital Wallet generated the highest sales for Electric household essentials, with 78,869.0.

3)Credit Card generated the highest sales for Butchers, with 75,545.0 — slightly higher than Cash.

4)Milk Products had the lowest sales across all payment methods, with values ranging from 61,469.5 to 63,209.0.

5)The highest individual value in the entire Pivot Table is 78,869.0, for Electric household essentials using Digital Wallet.
